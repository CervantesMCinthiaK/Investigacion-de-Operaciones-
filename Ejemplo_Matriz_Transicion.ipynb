Tomemos: 1 para Sol; 0 para Águila.

```python
def volado():
    v = r.random()
    if v <= 0.5:
       x = 1  # 1 para Sol
    else:
        x = 0  # 0 para Águila
    return x
```

```python
volado()
```

```python
def volado(n):
    lista = []
    for i in range(n):
        v = r.random()
        if v <= 0.5:
            x = 1
        else:
            x = 0
        lista.append(x)
    return lista
```

```python
# Ejemplo de uso (simulando 20 volados):
volado(20)
```
