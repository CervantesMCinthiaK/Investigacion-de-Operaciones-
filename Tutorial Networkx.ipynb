```python
import networkx as nx
import matplotlib.pyplot as plt
G = nx.Graph([(1, 2), (1, 3), (1, 5), (2, 3), (3, 4), (4, 5)])

# Explicitly set positions
pos = {1: (0, 0), 2: (-1, 0.3), 3: (2, 0.17), 4: (4, 0.255), 5: (5, 0.03)}

options = {
    "font_size": 36,
    "node_size": 3000,
    "node_color": "white",
    "edgecolors": "black",
    "linewidths": 5,
    "width": 5,
}

nx.draw(G, pos, with_labels=True, **options)
plt.show()
```

#Tarea


```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph([
    (1, 2), (1, 3), (3, 2),
    (2, 6), (6, 2),
    (2, 4), (3, 4), (4, 3),
    (2, 5), (5, 4),
    (4, 6), (3, 5), (5, 6),
    (4, 7), (6, 7), (5, 7)
])

# Explicitly set positions
pos = {
    1: (0, 1.5),
    2: (2, 3),
    3: (2, 0),
    4: (4, 1.5),
    5: (6, 0),
    6: (6, 3),
    7: (8, 1.5)
}

options = {
    "font_size": 16,
    "node_size": 1200,
    "node_color": "white",
    "edgecolors": "black",
    "linewidths": 2,
    "width": 2,
    "arrowsize": 20
}

nx.draw(G, pos, with_labels=True, **options)

labels = {
    (1, 2): 5,
    (1, 3): 1,
    (3, 2): 2,
    (2, 6): 6,
    (6, 2): 7,
    (2, 4): 7,
    (3, 4): 6,
    (4, 3): 7,
    (2, 5): 1,
    (5, 4): 3,
    (4, 6): 4,
    (3, 5): 7,
    (5, 6): 5,
    (4, 7): 6,
    (6, 7): 2,
    (5, 7): 9
}

nx.draw_networkx_edge_labels(G, pos, edge_labels=labels)

plt.show()
```
